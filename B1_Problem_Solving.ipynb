```python
attempts = int(input())
level = int(input()) # 1 = guest, 2 = user, 3 = admin
```

```python
if attempts > 5:
  print("Security Alert")
elif attempts > 3 and level == 1:
  print("Access Denied")
elif attempts <= 3 and level == 3:
  print("Welcome, Admin")
elif attempts == 0 and (level == 2 or level == 3):
  print("New Session")
```

```python
for i in range(1,4):
  for j in range(1,4):
  if i == j:
    continue
  elif i*j == 4:
    pass
  else:
    break
  print(i, '*', '=', i*j)
```

```python
for i in range(10, 0, -1):
  if i == 1:
    print(i)
  else:
    print(i, end='->')
```

```python
for i in range(1,4):
  for j in range(1,4):
    if i == j:
      continue
    elif i * j == 4:
      pass
    else:
      break
    print(i, '*', j, '=', i * j)
```

```python
def process_data(a, b=10):
  return a*b

x = 5
y = 20

result1 = process_data(x)
result2 = process_data(a=y)
result3 = process_data(a=x, b=y)

print(result1, result2, result3)
```
